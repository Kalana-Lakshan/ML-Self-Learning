```python
features = train_data.columns.tolist()
features
```

```python
features = features[1:-1]
features
```

```python
len(features)
```

```python
from sklearn.preprocessing import OrdinalEncoder
import numpy as np

features_to_encode = ['Chest pain type','EKG results','Slope of ST','Thallium']

train_data_to_encode = train_data[features_to_encode].copy()

for column in features_to_encode:
  if train_data_to_encode[column].isnull().any():
    mode_value = train_data_to_encode[column].mode()[0]
    train_data_to_encode[column].fillna(mode_value,inplace = True)

ordinalencoder = OrdinalEncoder()
train_data_to_encode[features_to_encode] = ordinalencoder.fit_transform(train_data_to_encode[features_to_encode])

train_data[features_to_encode] = train_data_to_encode[features_to_encode]



```

```python
test_data_to_encode = test_data[features_to_encode].copy()

for column in features_to_encode:
  if test_data_to_encode[column].isnull().any():
    mode_value = test_data_to_encode[column].mode()[0]
    test_data_to_encode[column].fillna(mode_value,inplace = True)

test_data[features_to_encode] = ordinalencoder.fit_transform(test_data_to_encode[features_to_encode])
```

```python
from sklearn.model_selection import train_test_split

X_train,X_val,y_train,y_val = train_test_split(train_data[features],train_data['Heart Disease'],test_size = 0.2,random_state = 42)

import xgboost as xgb

xgmodel = xgb.XGBClassifier(random_state = 42)
xgmodel.fit(X_train,y_train)

from sklearn.metrics import roc_auc_score

y_predict_probabilities = xgmodel.predict_proba(X_val)[:,1]

auc_roc_value = roc_auc_score(y_val,y_predict_probabilities)
```
